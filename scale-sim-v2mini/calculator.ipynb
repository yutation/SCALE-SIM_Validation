```python
from enum import Enum
import math
from typing import Tuple


class SystolicArrayDataflow(Enum):
    OS = 'os'
    WS = 'ws'
    IS = 'is'

class MatrixMatiplication:
    def __init__(self, m, n, k):
        #(m, n) * (n, k) = (m, k)
        self.m = m
        self.n = n
        self.k = k
        self.mp = 1
        self.np = 1
        self.kp = 1
        self.partition_num = self.mp * self.np * self.kp
        self.partition_m = math.ceil(self.m / self.mp)
        self.partition_n = math.ceil(self.n / self.np)
        self.partition_k = math.ceil(self.k / self.kp)

    def set_partition_num(self, mp, np, kp):
        self.mp = mp
        self.np = np
        self.kp = kp
        self.partition_num = self.mp * self.np * self.kp
        self.partition_m = math.ceil(self.m / self.mp)
        self.partition_n = math.ceil(self.n / self.np)
        self.partition_k = math.ceil(self.k / self.kp)

    def set_partition_size(self, m, n, k):
        self.partition_m = m
        self.partition_n = n
        self.partition_k = k
        self.mp = math.ceil(self.m / self.partition_m)
        self.np = math.ceil(self.n / self.partition_n)
        self.kp = math.ceil(self.k / self.partition_k)
        self.partition_num = self.mp * self.np * self.kp

class SystolicArray:
    def __init__(self, rows, cols, dataflow: SystolicArrayDataflow):
        self.rows = rows
        self.cols = cols
        self.dataflow: SystolicArrayDataflow = dataflow

    def matrix_multiply(self, m, n, k):
        sr = 0
        sc = 0
        t = 0
        if self.dataflow == SystolicArrayDataflow.OS:
            sr = m
            sc = n
            t = k
        elif self.dataflow == SystolicArrayDataflow.WS:
            sr = k
            sc = n
            t = m
        elif self.dataflow == SystolicArrayDataflow.IS:
            sr = k
            sc = m
            t = n
        else:
            raise ValueError(f"Invalid dataflow: {self.dataflow}")

        total_cycles = (2 * self.rows + self.cols + t - 2) * math.ceil(sr / self.rows) * math.ceil(sc / self.cols)
        return total_cycles

class SystolicArrayParallel:
    def __init__(self, rows, cols, dataflow: SystolicArrayDataflow):
        self.rows = rows
        self.cols = cols
        self.dataflow: SystolicArrayDataflow = dataflow

    def matrix_multiply(self, matmul: Tuple[int, int, int], partition: Tuple[int, int, int]):
        m, n, k = matmul
        mp, np, kp = partition
        m = math.ceil(m / mp)
        n = math.ceil(n / np)
        k = math.ceil(k / kp)
        if self.dataflow == SystolicArrayDataflow.OS:
            sr = m
            sc = n
            t = k
        elif self.dataflow == SystolicArrayDataflow.WS:
            sr = k
            sc = n
            t = m
        elif self.dataflow == SystolicArrayDataflow.IS:
            sr = k
            sc = m
            t = n
        else:
            raise ValueError(f"Invalid dataflow: {self.dataflow}")

        total_cycles = (2 * self.rows + self.cols + t - 2) * math.ceil(sr / self.rows) * math.ceil(sc / self.cols)
        return total_cycles


class TPU():
    def __init__(self, rows = 128, cols = 128, num = 8):
        self.rows = rows
        self.cols = cols
        self.num = num
        self.systolic_array_ws = SystolicArray(rows, cols, SystolicArrayDataflow.WS)
        self.systolic_array_is = SystolicArray(rows, cols, SystolicArrayDataflow.IS)

    def schedule(self, matmul: MatrixMatiplication):
        pass


    def run_one_matrix_multiplication(self, matmul: MatrixMatiplication):
        return super().run_one_matrix_multiplication(matmul)
    


```

```python
systolic_array = SystolicArray(128, 128, SystolicArrayDataflow.WS)
matmul = MatrixMatiplication(512, 512, 512)
print(systolic_array.run_one_matrix_multiplication(matmul))
```
